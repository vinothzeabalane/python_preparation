Inheritance is an object-oriented programming concept where a child class acquires the attributes and methods of a parent class.

It allows us to:
- Reuse existing code.
- Extend parent-class functionality.
- Override parent methods.
- Organize related classes into a hierarchy.

single Inheritance

```python
class Parent:
    def display(self):
        print("This is the parent class.")


class Child(Parent):
    def display(self):
        super().display()
        print("This is the child class.")

child_object = Child()
child_object.display()  # Output: This is the parent class. This is the child class.
```

multiple Inheritance

```python
class Parent1:
    def display(self):
        print("This is the parent class 1.")    


class Parent2:
    def display(self):
        print("This is the parent class 2.")


class Child(Parent2, Parent1):
    def display(self):
        super().display()
        print("This is the child class.")

child_object = Child()
child_object.display()  # Output: This is the parent class 1. This is the parent class 2. This is the child class.
```

multilevel Inheritance

```python
class Grandparent:
    def display(self):
        print("This is the grandparent class.") 

class Parent(Grandparent):
    def display(self):
        super().display()
        print("This is the parent class.")

class Child(Parent):
    def display(self):
        super().display()
        print("This is the child class.")   

child_object = Child()
child_object.display()  # Output: This is the grandparent class. This is the parent class. This is the child class.
```

Hierarchical Inheritance

```python
class Parent:
    def display(self):
        print("This is the parent class.")  

class Child1(Parent):
    def display(self):
        super().display()
        print("This is the child class 1.")

class Child2(Parent):
    def display(self):
        super().display()
        print("This is the child class 2.") 

child1_object = Child1()
child1_object.display()  # Output: This is the parent class. This is the child class 1.
child2_object = Child2()
child2_object.display()  # Output: This is the parent class. This is the child class 2.
```

Hybrid Inheritance

```python
class Grandparent:
    def display(self):
        print("This is the grandparent class.") 

class Parent1(Grandparent):
    def display(self):
        print("This is the parent class 1.")    

class Parent2(Grandparent):
    def display(self):
        print("This is the parent class 2.")    

class Child(Parent1, Parent2):
    
    def display(self):
        super().display()
        print("This is the child class.")

child_object = Child()
child_object.display()  # Output: This is the grandparent class. This is the parent class 1. This is the parent class 2. This is the child class.
```
