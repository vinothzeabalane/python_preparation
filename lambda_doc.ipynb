 A lambda function is a small, anonymous function that is defined using the lambda keyword. It can accept multiple arguments but contains only one expression.
- lambda – defines the function.
- Arguments – inputs passed to the function.
- Expression – operation performed and returned automatically.
- No explicit return statement is required.
    

lambda arguments: expression

Type1: No argument

```python
res = lambda: "No argument"
print(res())
```

Positional argument

```python
res = lambda x: x ** 2
print(res(5))
```

```python
res = lambda x, y: x + y
print(res(5, 10))
```

```python
res = lambda *args: sum(args)
print(res(1, 2, 3, 4, 5))
```

```python
res = lambda **kwargs: kwargs
print(res(a=1, b=2, c=3))   
```

```python
max_value = lambda x, y: x if x > y else y
print(max_value(10, 20))
```

```python
numbers = [1, 2, 3, 4, 5]   
result = list(map(lambda x: x * 2, numbers))
print(result)
```

```python
filtered_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(filtered_numbers)
```

```python
from functools import reduce

reduced_sum = reduce(lambda x, y: x + y, numbers)
print(reduced_sum)
```

```python
people = [{"name": "Ada", "age": 36}, {"name": "Linus", "age": 54}, {"name": "Grace", "age": 85}, {"name": "Alan", "age": 42}]
by_age = sorted(people, key=lambda person: person["age"])
print(by_age)
```
